Напишем функцию, которая принимает на вход название целочисленного признака и количество бинов, а выдаёт три графика:

Теперь апишем функцю, которая принимает на вход название категориального признака и количество бинов, а выдаёт три графика:
- Счётную гистограмму признака на тренировочной выборке
- Счётную гистограмму признака на тестовой выборке
- Счётную гистограмму признака на тренировочной выборке в разрезе целевой переменной

```python
def plot_count_histogram(column, hue='Churn', figsize=(18, 7), rotation=0):
    _, axes = plt.subplots(nrows=1, ncols=3, figsize=figsize)
    sns.countplot(data=train, x=column, color='blue', ax=axes[0])
    axes[0].bar_label(axes[0].containers[0])
    sns.countplot(data=test, x=column, color='orange', ax=axes[1])
    axes[1].bar_label(axes[1].containers[0])
    sns.countplot(data=train, x=column, hue='Churn', ax=axes[2])
    for i in range(2):
        axes[2].bar_label(axes[2].containers[i])
    for ax in axes:
        for label in ax.get_xticklabels():
            label.set_rotation(rotation)
            label.set_ha('right')
    plt.show()
```

Постром счётную диаграмму для признака пола

```python
plot_count_histogram('Sex')
```

Выводы:
- Распределение на тестовой выборке имеет чуть больший перекос в сторону мужчин, чем на тренировочной
- В разрезе целевой переменной оба распределния выглядят одинаково, то есть по признаку полу всегда выделяется ~35-36% ушедших клиентов

## Приготовления к обучению

Подбирать оптимальные гиперпараметры для моделй будем с помощью фреймворка Optuna, для этого напишем функцию search, которая принимает на вход модель, ловарь с гиперпараметрами, количество итераций, тренировочную выборку и целевую переменную, а вовзаращет экземпляр обучения, который наилучшим образом показал себя на кросс-валидации

```python
def search(estimator, distributions, n_trials, X_train, y_train):
    optuna_search = OptunaSearchCV(
        estimator=estimator,
        param_distributions=distributions,
        cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=42),
        n_trials=n_trials, scoring='roc_auc',
        verbose=1
    )
    optuna_search.fit(X_train, y_train)
    return optuna_search
```

Теперь напишем функцию для отрисовки графика кривой ROC-AUC для тренировочной и тестовой выборок

```python
def plot_roc_auc_curve(train_proba, test_proba, y_train, y_test, title):
    fpr, tpr, _ = roc_curve(y_train, train_proba)
    plt.plot(fpr, tpr, color='blue', label=f'train ROC-AUC = {roc_auc_score(y_train, train_proba):.5f}')

    fpr, tpr, _ = roc_curve(y_test, test_proba)
    plt.plot(fpr, tpr, color='orange', label=f'test ROC-AUC = {roc_auc_score(y_test, test_proba):.5f}')

    plt.plot([0, 1], [0, 1], color='black', linestyle='--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(title)
    plt.legend()
    plt.show()
```

Замечаем, что если значением признака наличия интернета является No, то у признаков наличия онлайн-сервиса безопасности, онлайн-бэкапов, защиты девайсов, доступа к техподдержке, онлайн-телевидения и подписки на онлайн-кинотеатр будет одно и тоже значение No Internet service

Отсюда рождается идея, резделить тренировочную и тетстовую выборку на две части - в одной будут те, у кого еть доступ в интернет, а в другой те, у кого такого доступа нет. При этом для "оффлайн"-пользователей следует удалить вышеназванные признаки с одинаковым значением

Напишем функцию для кодировки категориальных и нормирования всех признаковых столбцов у тренировочной и тестовой выборки

```python
def encode_and_scale(train, test):
    encoder = CatBoostEncoder()
    train_scaled, test_scaled = train.copy(), test.copy()
    train_scaled.loc[:, cat_cols] = encoder.fit_transform(train[cat_cols], train['Churn'])
    test_scaled.loc[:, cat_cols] = encoder.transform(test[cat_cols])
    scaler = StandardScaler()
    train_scaled.loc[:, feature_cols] = scaler.fit_transform(train_scaled[feature_cols])
    test_scaled.loc[:, feature_cols] = scaler.transform(test_scaled[feature_cols])
    return train_scaled, test_scaled
```

Напишем функцию для получения общего предсказания из двух моделей

```python
def get_probs(model_internet, model_no_internet, df, df_internet, df_no_internet, name='Churn'):
    proba = pd.Series(index=df.index, name=name)
    proba.loc[df_internet.index] = model_internet.predict_proba(df_internet)[:, 1]
    proba.loc[df_no_internet.index] = model_no_internet.predict_proba(df_no_internet)[:, 1]
    return proba
```

**Внимание!**
Так как мы будем обучать две разные модели для двух тренировочных выборок, то для адекватного значения метрики ROC-AUC для всей тренировочной выборки целиком (вероятности ведь будут предсказываться двумя разными моделями), следует использовать калибровку вероятностей, она сможет в какой-то степени невелировать разницу между предсказаниями моделей. ## Кластеризация

Используем кластеризацию для создания новых признаков - номеров кластеров, для этого создадим датафреймы для тренировочной и тестовой выборки

```python
cluster_cols = []
train_clusters = pd.DataFrame(index=train.index)
test_clusters = pd.DataFrame(index=test.index)
```

Теперь напишем функцию, которая принимает на вход двумерные тренировочную и тестовую выборки для визуализации, и выводит 2 графика:
- Точки тренировочной выборки, окрашенные в разные цвета в зависимости от класса
- Точки, окрашенные в разные цвета в зависимости от того, относится объект к тренировочным или к тестовым данным

```python
def plot_clusters(train_2d, test_2d, train, figsize=(18, 7)):
    _, axes = plt.subplots(nrows=1, ncols=2, figsize=figsize)
    axes[0].scatter(train_2d.loc[train[train['Churn'] == 0].index, 0], train_2d.loc[train[train['Churn'] == 0].index, 1], color='blue', label='0')
    axes[0].scatter(train_2d.loc[train[train['Churn'] == 1].index, 0], train_2d.loc[train[train['Churn'] == 1].index, 1], color='red', label='1')
    axes[0].set_title('Двумерные точки train в разрезе класса')
    axes[1].scatter(train_2d.loc[:, 0], train_2d.loc[:, 1], color='green', label='Train')
    axes[1].scatter(test_2d.loc[:, 0], test_2d.loc[:, 1], color='orange', label='Test')
    axes[1].set_title('Двумерные точки train и test')
    for ax in axes:
        ax.legend()
    plt.show()
```

Дальше напишем функцию для отрисовки предсказаний кластеров для тренировочной и тестовой выборок

```python
def plot_preds(train_2d, train_preds, test_2d, test_preds, figsize=(18, 7)):
    _, axes = plt.subplots(nrows=1, ncols=2, figsize=figsize)
    sns.scatterplot(x=train_2d.loc[:, 0], y=train_2d.loc[:, 1], hue=train_preds, palette='viridis', legend='full', ax=axes[0])
    axes[0].set_title('Train')
    sns.scatterplot(x=test_2d.loc[:, 0], y=test_2d.loc[:, 1], hue=test_preds, palette='viridis', legend='full', ax=axes[1])
    axes[1].set_title('Test')
    for ax in axes:
        ax.legend()
    plt.show()
```

Напишем функцию перенумерации кластеров так, чтобы чем больше был клатер, тем больший порядковый номер ему бы соответствовал

```python
def reindex(preds):
    statistics = preds.value_counts()
    current_index = statistics.index
    sorted_index = statistics.sort_index(ascending=False).index
    preds = preds.map(dict(zip(current_index, sorted_index)))
    return preds
```

# Предсказания

Для начала посчитаем значение метрики ROC-AUC для трёх моделей градиентного бустинга, однако так как моделей две, то я будем использовать взвешенное среднее из коэффициентов

```python
def get_coefficient(score_internet, score_no_internet, df_internet, df_no_internet):
    return (score_internet * df_internet.shape[0] + score_no_internet * df_no_internet.shape[0]) / (df_internet.shape[0] + df_no_internet.shape[0])
```

```python
scores = {
    'catboost': get_coefficient(
        best_cb_internet.best_score_, best_cb_no_internet.best_score_,
        X_train_internet, X_train_no_internet
    ),
    'XGBoost': get_coefficient(
        best_xgb_internet.best_score_, best_xgb_no_internet.best_score_,
        X_train_internet, X_train_no_internet
    ),
    'LightGBM': get_coefficient(
        best_lgbm_internet.best_score_, best_lgbm_no_internet.best_score_,
        X_train_internet, X_train_no_internet
    )
}
```

```python
scores
```

Теперь нормируем каждое значение на сумму метрик ROC-AUC

```python
coefs = {}
denominator = sum(scores.values())
for key, value in scores.items():
    coefs[key] = value / denominator
```

Предскажем вероятности каждой моделью

```python
sample_submission = pd.DataFrame({
    'CatBoost': cb_test_probs.values,
    'XGBoost': xgb_test_probs.values,
    'LightGBM': lgbm_test_probs.values
}, index=test.index)
sample_submission.index.name = 'Id'
```

```python
sample_submission
```

Теперь умножим предсказние каждой модели на свой коэффициент и сложим, так и получается финальная вероятность

```python
submission = pd.DataFrame((sample_submission * coefs.values()).sum(axis=1), columns=['Churn'])
submission.head()
```
